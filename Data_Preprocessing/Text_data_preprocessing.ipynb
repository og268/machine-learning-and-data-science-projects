Three of the letters in the variable book have been swapped for punctuation: each 'h' has been replaced with '{', each 'a' has been replaced with '@' and each 'o' has been replaced with '['.

Below I swap the punctuation for the relevant letter


```python
punctuation_mappings = {"{":"h", "@":"a", "[":"o"}

clean_book = book

for symbol, letter in punctuation_mappings.items():
    clean_book = clean_book.replace(symbol, letter)
    
print(clean_book)
```

Next I remove all punctuation from clean_book.

```python
import string

clean_book_no_punc = clean_book

for char in string.punctuation:
    clean_book_no_punc = clean_book_no_punc.replace(char, '')
    
    
print(clean_book_no_punc)
```

Next I convert this into a list of words

```python
word_list = clean_book_no_punc.split()
word_list
```

Find the longest word - defined by number of characters - in word_list.

```python
longest_word = max(word_list, key=lambda x: len(x))
longest_word
```

I now construct a dictionary, word_freqs, where each key is the length of a word and each value is the count of words of that length in the list.


```python
def createDicFromWordList(word_list):
    word_freqs = {}
    for word in word_list:
        word_length = len(word)
        if word_length in word_freqs:
            word_freqs[word_length] += 1
        else: 
            word_freqs[word_length] = 1
    return word_freqs

word_freqs = createDicFromWordList(word_list)
word_freqs
```

```python
# ['quick', 'brown', 'fox'] should map to {5: 2, 3: 1}.
test_list = ['quick', 'brown', 'fox']

createDicFromWordList(test_list)
```

Find the most common word length

```python
most_common_word_length = max(word_freqs, key=word_freqs.get)
most_common_word_length
```

Next I construct a dictionary, letter_freqs, using clean_book_no_punc, where each key is a letter and each value is the frequency of that letter in the string. Uppercase and lowercase letters must both be included in the letter count. The lowercase letter should be used as the key.
For example 'The start' would have an entry {'t':3} in the letter_freqs dictionary. This should not include special characters (e.g. spaces or newlines).**


```python
clean_book_no_punc
```

```python
lowercase_letters = string.ascii_lowercase
lowercase_letters
```

```python
def returnLetterFreqs(string:str) -> dict:
    letter_freqs = {}
    for letter in string.lower():
        if letter in lowercase_letters:
            if letter in letter_freqs:
                letter_freqs[letter] += 1
            else:
                letter_freqs[letter] = 1
                
    return letter_freqs
```

```python
returnLetterFreqs(clean_book_no_punc)
```
